```python
# Write a program to add a key and value to a dictionary.

dictionary = {0: 10, 1: 20}

dictionary[2] = 30

print(dictionary)
```

```python
# Write a program to concatenate the following dictionaries to create a new one.

dic1 = {1: 10, 2: 20}
dic2 = {3: 30, 4: 40}
dic3 = {5: 50, 6: 60}

dic4 = dic1.copy()
dic4.update(dic2)
dic4.update(dic3)

print(dic4)
```

```python
# Write a program to check if a given key already exists in a dictionary.

dictionary = {1: 10, 2: 20, 3: 30, 4: 40}

key = int(input("Enter the key: "))

if key in dictionary:
    print("Key already exists")
else:
    print("Key does not exist")
```

```python
# Write a program to iterate over a dictionary using for loop and print the keys alone, values alone and both keys and values.

dictionary = {1: 10, 2: 20, 3: 30, 4: 40}

print("Keys:")
for key in dictionary:
    print(key)

print("Values:")
for value in dictionary.values():
    print(value)

print("Keys and Values:")
for key, value in dictionary.items():
    print(key, value)
```

```python
# Write a program to prepare a dictionary where the keys are numbers
# between 1 and 15 (both included) and the values are square of the keys.

dictionary = {}

for i in range(1, 16):
    dictionary[i] = i ** 2

print(dictionary)
```

```python
# Write a program to sum all the values in a dictionary,
# considering the values will be of int type.

dictionary = {1: 10, 2: 20, 3: 30, 4: 40}

total = 0

for value in dictionary.values():
    total = total + value

print("Sum of all values:", total)
```

```python
# Assign 5 elements in dictionary named my_dict.This dictionary contains student's information.
# Add and delete student information from my_dict, Display student information.

my_dict = {
    "Name": "Rahul",
    "Roll No": 101,
    "Age": 18,
    "Course": "CSE",
    "College": "ABC College"
}

# Add student information
my_dict["Marks"] = 85

# Delete student information
del my_dict["Age"]

# Display student information
print("Student Information:")

for key, value in my_dict.items():
    print(key, ":", value)
```

```python
# Convert list to dictionary.

list1 = ["name", "panel", "rollno"]
list2 = ["ABC", "B", 34]

my_dict = dict(zip(list1, list2))

print(my_dict)
```

```python
# Convert dictionary to lists. Make one list of keys and other of values.

my_dict = {
    "name": "ABC",
    "panel": "B",
    "rollno": 34
}

keys = list(my_dict.keys())
values = list(my_dict.values())

print("List of keys:", keys)
print("List of values:", values)
```

```python
# Find the mean of values of dictionary.

mydict = {
    "marks1": 23,
    "marks2": 123,
    "marks3": 43,
    "marks4": 13,
    "marks5": 39
}

total = sum(mydict.values())
mean = total / len(mydict)

print("Mean of all values:", mean)
```

```python
# Sort the elements of the dictionary using sorted() function.

my_dict = {
    "name": "ABC",
    "panel": "B",
    "rollno": 34,
    "marks": [65, 87, 67, 94]
}

print("Sorted dictionary keys:", sorted(my_dict))
```
