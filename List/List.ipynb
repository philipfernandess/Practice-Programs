```python
# Accept 20 values from user and save it in list.
# Perform following operations on it
# a) Count similar elements of list and print their index value
# b) Count even and odd values of list
# c) Count positive and negative values of list

numbers = []

for i in range(20):
    num = int(input("Enter a number: "))
    numbers.append(num)

print("List:", numbers)

# a) Count similar elements and print their index values

for i in range(20):
    if numbers.count(numbers[i]) > 1:
        if numbers.index(numbers[i]) == i:
            print(numbers[i], "occurs", numbers.count(numbers[i]), "times")
            print("Index:", end=" ")

            for j in range(20):
                if numbers[j] == numbers[i]:
                    print(j, end=" ")

            print()

# b) Count even and odd values

even = 0
odd = 0

for num in numbers:
    if num % 2 == 0:
        even = even + 1
    else:
        odd = odd + 1

print("Even values:", even)
print("Odd values:", odd)

# c) Count positive and negative values

positive = 0
negative = 0

for num in numbers:
    if num > 0:
        positive = positive + 1
    elif num < 0:
        negative = negative + 1

print("Positive values:", positive)
print("Negative values:", negative)
```

```python
# Accept 10 values from user and save it in list.
# Perform following operations on it
# a) Sort list in ascending order using sorted() function and display sorted list
# b) Sort list in descending order using sort() function
# c) Display length of list

numbers = []

for i in range(10):
    num = int(input("Enter a number: "))
    numbers.append(num)

# a) Sort list in ascending order using sorted()

ascending = sorted(numbers)
print("Ascending order:", ascending)

# b) Sort list in descending order using sort()

numbers.sort(reverse=True)
print("Descending order:", numbers)

# c) Display length of list

print("Length of list:", len(numbers))
```

```python
# Accept two lists from user and merge them using + in a single list

list1 = []
list2 = []

print("Enter 5 values for first list:")
for i in range(5):
    num = int(input("Enter a number: "))
    list1.append(num)

print("Enter 5 values for second list:")
for i in range(5):
    num = int(input("Enter a number: "))
    list2.append(num)

list3 = list1 + list2

print("First list:", list1)
print("Second list:", list2)
print("Merged list:", list3)
```

```python
# An acronym is a word formed by taking the first letters of the words in a phrase.
# Write a program that allows the user to type in a phrase and then outputs
# the acronym for that phrase. The acronym should be in uppercase.

phrase = input("Enter a phrase: ")

words = phrase.split()
acronym = ""

for word in words:
    acronym = acronym + word[0]

print("Acronym:", acronym.upper())
```

```python
# Write a program to print the abbreviation of a month, given its number.

month = int(input("Enter month number: "))

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

if month >= 1 and month <= 12:
    print("Month number:", month)
    print("Month abbreviation:", months[month - 1])
else:
    print("Invalid month number")
```
