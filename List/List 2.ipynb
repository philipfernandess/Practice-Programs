```python
# 9. Accept 20 values from user and save it in list. Perform following operations:
# a) Count similar elements of list and print their index value
# b) Count even and odd values of list
# c) Count positive and negative values of list

L = []

for i in range(20):
    num = int(input("Enter a number: "))
    L.append(num)

print("List =", L)

# a) Count similar elements and print their index values

for x in L:
    if L.count(x) > 1:
        print(x, "occurs", L.count(x), "times")
        print("Index values:", end=" ")

        for i in range(20):
            if L[i] == x:
                print(i, end=" ")

        print()

# b) Count even and odd values

even = 0
odd = 0

for x in L:
    if x % 2 == 0:
        even = even + 1
    else:
        odd = odd + 1

print("Even values =", even)
print("Odd values =", odd)

# c) Count positive and negative values

positive = 0
negative = 0

for x in L:
    if x > 0:
        positive = positive + 1
    elif x < 0:
        negative = negative + 1

print("Positive values =", positive)
print("Negative values =", negative)
```

```python
# 10. Accept 10 values from user and save it in list. Perform following operations:
# a) Sort list in ascending order using sorted() function and display sorted list
# b) Sort list in descending order using sort() function
# c) Display length of list

L = []

for i in range(10):
    num = int(input("Enter a number: "))
    L.append(num)

# a) Ascending order using sorted()

ascending = sorted(L)

print("Ascending order =", ascending)

# b) Descending order using sort()

L.sort(reverse=True)

print("Descending order =", L)

# c) Display length

print("Length of list =", len(L))
```

```python
# 12. An acronym is a word formed by taking the first letters of the words
# in a phrase and making a word from them.
# Write a program that allows the user to type in a phrase and then outputs
# the acronym for that phrase.
# The acronym should be all uppercase.

phrase = input("Enter a phrase: ")

words = phrase.split()

acronym = ""

for word in words:
    acronym = acronym + word[0]

print("Acronym =", acronym.upper())
```

```python
# 13. Write a program to print the abbreviation of a month, given its number.

months = [
    "Jan", "Feb", "Mar", "Apr",
    "May", "Jun", "Jul", "Aug",
    "Sep", "Oct", "Nov", "Dec"
]

num = int(input("Enter month number: "))

if num >= 1 and num <= 12:
    print("Month abbreviation =", months[num - 1])
else:
    print("Invalid month number")
```

```python
# 15. Python program to merge two lists into a dictionary
# (one list as keys, another as values).

keys = ["name", "age", "city"]
values = ["ABC", 20, "Pune"]

my_dict = dict(zip(keys, values))

print("Dictionary =", my_dict)
```

```python
# 16. Python program to remove duplicate elements from a list.

L = [10, 20, 10, 30, 20, 40, 30]

new_list = []

for x in L:
    if x not in new_list:
        new_list.append(x)

print("Original list =", L)
print("List after removing duplicates =", new_list)
```

```python
# 18. Python program to find common elements in two lists.

list1 = [10, 20, 30, 40, 50]
list2 = [30, 40, 50, 60, 70]

common = []

for x in list1:
    if x in list2:
        common.append(x)

print("Common elements =", common)
```
