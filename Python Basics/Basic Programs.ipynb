```python
# 7. Write a program to print the sum of all the digits of a given number.

num = int(input("Enter a number: "))

sum = 0

while num > 0:
    digit = num % 10
    sum = sum + digit
    num = num // 10

print("Sum of digits =", sum)
```

```python
# 8. Write a program to find the factorial of a given number.

num = int(input("Enter a number: "))

fact = 1

for i in range(1, num + 1):
    fact = fact * i

print("Factorial =", fact)
```

```python
# 9. Write a Python program to check if a given number is prime or not.

num = int(input("Enter a number: "))

if num < 2:
    print("Not a prime number")
else:
    for i in range(2, num):
        if num % i == 0:
            print("Not a prime number")
            break
    else:
        print("Prime number")
```

```python
# 10. Write a Python program to print prime numbers between 10 to 99.

for num in range(10, 100):
    if num < 2:
        continue

    for i in range(2, num):
        if num % i == 0:
            break
    else:
        print(num)
```

```python
# 12. Write a program to reverse a given number.

num = int(input("Enter a number: "))

reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10

print("Reverse =", reverse)
```

```python
# 13. Write a program to determine whether a given year is a leap year or not.

year = int(input("Enter a year: "))

if year % 400 == 0:
    print("Leap year")
elif year % 100 == 0:
    print("Not a leap year")
elif year % 4 == 0:
    print("Leap year")
else:
    print("Not a leap year")
```

```python
# 14. In a certain university, grades are assigned to the marks obtained
# in respective subjects. Write a Python program to accept marks in Math
# and display the respective grade as an output using if...elif...else
# statements (using multiple conditions).
#
# Marks obtained    Grade assigned
# 90-100            O
# 80-89             A+
# 70-79             A
# 60-69             B
# 50-59             C
# 40-49             P
# Below 40          F

marks = int(input("Enter marks in Math: "))

if marks >= 90 and marks <= 100:
    print("Grade = O")
elif marks >= 80 and marks <= 89:
    print("Grade = A+")
elif marks >= 70 and marks <= 79:
    print("Grade = A")
elif marks >= 60 and marks <= 69:
    print("Grade = B")
elif marks >= 50 and marks <= 59:
    print("Grade = C")
elif marks >= 40 and marks <= 49:
    print("Grade = P")
else:
    print("Grade = F")
```

```python
# 16. Accept Student Name, Roll Number and Marks of the 3 subjects
# from the user. Calculate the percentage of the marks and display it.
# Display the subject with Highest and lowest marks.

name = input("Enter Student Name: ")
roll_no = input("Enter Roll Number: ")

math = float(input("Enter Math marks: "))
python = float(input("Enter Python marks: "))
science = float(input("Enter Science marks: "))

total = math + python + science
percentage = total / 3

print("Student Name =", name)
print("Roll Number =", roll_no)
print("Percentage =", percentage)

if math > python and math > science:
    print("Highest marks = Math")
elif python > math and python > science:
    print("Highest marks = Python")
else:
    print("Highest marks = Science")

if math < python and math < science:
    print("Lowest marks = Math")
elif python < math and python < science:
    print("Lowest marks = Python")
else:
    print("Lowest marks = Science")
```
