```python
# 1. Write a program to remove a given item from the set.

my_set = {10, 20, 30, 40, 50}

print("Original Set:", my_set)

item = int(input("Enter item to remove: "))

my_set.remove(item)

print("Set after removing item:", my_set)
```

```python
# 2. Write a program to create an intersection of sets.

set1 = {10, 20, 30, 40, 50}
set2 = {30, 40, 50, 60, 70}

intersection = set1.intersection(set2)

print("Set 1:", set1)
print("Set 2:", set2)
print("Intersection:", intersection)
```

```python
# 3. Write a program to create an union of sets.

set1 = {10, 20, 30, 40, 50}
set2 = {30, 40, 50, 60, 70}

union = set1.union(set2)

print("Set 1:", set1)
print("Set 2:", set2)
print("Union:", union)
```

```python
# 4. Write a program to find the maximum and minimum value in a set.

my_set = {25, 10, 45, 5, 60, 30}

maximum = max(my_set)
minimum = min(my_set)

print("Set:", my_set)
print("Maximum value:", maximum)
print("Minimum value:", minimum)
```
